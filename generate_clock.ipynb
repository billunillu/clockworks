# Generate a prayer-times clock image

Run the cells below to calculate the clock data and create `output/clock.svg` and `output/clock.png`. The defaults (location, timezone, and calculation method) are defined in `clock_image_generator.py`.

```python
from pathlib import Path
import importlib.util
import os
import subprocess
import sys
from IPython.display import Image, display

# Find this project from the current folder or one of its parent folders.
PROJECT_DIR = next(
    (folder for folder in (Path.cwd(), *Path.cwd().parents)
     if (folder / 'clock_image_generator.py').is_file()),
    None,
)
if PROJECT_DIR is None:
    raise FileNotFoundError(
        'Could not find clock_image_generator.py. Open this notebook from its project folder.'
    )
os.chdir(PROJECT_DIR)
sys.path.insert(0, str(PROJECT_DIR))

# Install project requirements into this notebook's active Python kernel if needed.
required_packages = ('adhanpy', 'hijridate', 'matplotlib', 'qrcode')
missing_packages = [name for name in required_packages if importlib.util.find_spec(name) is None]
if missing_packages:
    print(f'Installing missing packages: {", ".join(missing_packages)}')
    subprocess.check_call([
        sys.executable, '-m', 'pip', 'install', '-r', str(PROJECT_DIR / 'requirements.txt')
    ])

from clock_image_generator import get_chart_info, generate_image
print(f'Project: {PROJECT_DIR}')
```

```python
# Optional date. Leave DATE as None to use today in the configured timezone.
DATE = None                 # e.g. '2026-11-17' or '17-11-2026'

chart_info = get_chart_info(date=DATE)
chart_info['times']
```

```python
files = generate_image(chart_info)
print(f"SVG: {files['svg']}")
print(f"PNG: {files['png']}")
display(Image(filename=files['png']))
```
